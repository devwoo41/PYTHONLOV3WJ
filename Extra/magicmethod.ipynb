#### 매직 메서드
#### (= double underscore, Special method)

- 매직 메서드는 이름 앞뒤에 밑줄 두 개가 붙는 특별한 메서드다.

매직 메서드는 우리가 직접 호출하기보다는, 특정 문법이나 내장 함수를 사용할 때 파이썬 인터프리터가 자동으로 호출한다.

```python
print(obj)       # obj.__str__()
obj1 + obj2      # obj1.__add__(obj2)
obj[index]       # obj.__getitem__(index)
obj()            # obj.__call__()
```

```python
# 매직 메서드가 없는 경우, 사용자 정의 객체를 다룰 때마다 전용 메서드를 만들어야 한다.
class Wallet:
    def __init__(self, money):
        self.money = money

    def add_wallet(self, other):
        return Wallet(self.money + other.money)

    def is_same_money(self, other):
        return self.money == other.money
```

```python
wallet1 = Wallet(1000)
wallet2 = Wallet(2000)

wallet3 = wallet1.add_wallet(wallet2)
print(wallet3.money)  
wallet1.is_same_money(wallet2)  
```

```python
# 매직 메서드가 있는 경우, 연산자 오버로딩을 통해 사용자 정의 객체를 다룰 수 있다.
class Wallet:
    def __init__(self, money):
        self.money = money

    def __add__(self, other):
        return Wallet(self.money + other.money)

    def __eq__(self, other):
        return self.money == other.money
```

```python
wallet1 = Wallet(1000)
wallet2 = Wallet(2000)
wallet3 = wallet1 + wallet2
print(wallet3.money)
print(wallet1 == wallet2)
```

```python
# __init__ : 객체가 생성될 때 호출되는 메서드, 객체의 초기화를 담당
class User:
    def __init__(self, name, age):
        # 이미 생성된 객체의 초기상태를 설정
        self.name = name
        self.age = age

user = User("우주", 23)
print(user.name)  
print(user.age)
```

```python
# __new__ : 객체가 생성될 때 호출되는 메서드, 객체의 생성을 담당
class User:
    def __new__(cls, name):
        print("__new__: 객체 생성")

        # 실제 객체를 생성해서 반화한다.
        instance = super().__new__(cls)
        return instance

    def __init__(self, name):
        print("__init__: 객체 초기화")
        self.name = name

user = User("우주")
print(user.name) 
```

```python
# __str__ : 객체를 문자열로 표현할 때 호출되는 메서드, print() 함수나 str() 함수를 사용할 때 호출
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        # 반드시 문자열을 반환해야 한다.
        return f"{self.name}, {self.age}세"

user = User("우주", 23)

print(user)  # user.__str__() 호출
print(str(user))  # user.__str__() 호출
```

```python
# __repr__ : 객체의 상태를 확인하기 좋은 표현 정의 메서드, 객체를 표현할 때 호출, repr() 함수를 사용할 때 호출
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        # 반드시 문자열을 반환해야 한다.
        return f"User(name={self.name!r}, age={self.age!r})"

user = User("우주", 23)

print(repr(user))  # user.__repr__() 호출
```

이와 같이 매직 메서드는 파이썬의 문법과 내장 함수가 사용자 정의 객체와 대화하는 약속으로 매직 메서드는 여러 종류가 있다.

##### 메모리 주소를 가져오지 않으면서 복사하기 1 - [:] 활용

```python
a = [1, 2, 3]
b = a[:]
print(a, b)
```

```python
a[1] = 4
print(a, b)    
```

```python
a is b
```

메모리 주소를 가져오지 않으면서 복사하기 2 - copy 모듈 활용

```python
from copy import copy
a = [1, 2, 3]
b = copy(a)
print(a, b)
```

```python
a[1] = 4
print(a, b)
```

```python
a is b
```

##### 변수를 만드는 여러 가지 방법

```python
a, b = ('python', 'life')
print(a, b)
```

```python
(a, b) = 'python', 'life'
print(a, b)
```

```python
a, b = ['python', 'life']
print(a, b)
```

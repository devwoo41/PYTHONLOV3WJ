```python
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
```

```python
s1 & s2  # 교집합
```

```python
s1.intersection(s2)  # 교집합
```

```python
s1 | s2  # 합집합
```

```python
s1.union(s2)  # 합집합
```

```python
s1 - s2  # 차집합
```

```python
s1.difference(s2)  # 차집합
```

```python
s1 = set([1, 2, 3])
s1.add(4)
s1
```

```python
s1 = set([1, 2, 3])
s1.update([4, 5, 6])
s1
```

```python
s1 = set([1, 2, 3])
s1.remove(2)
s1
```

```python
s1 = set([1, 2, 3])
s1.discard(4)  # 없는 값을 제거하려고 하면 에러가 발생하지 않음
s1
```

```python
s1 = set([1, 2, 3])
s1.clear()  # 모든 요소 제거
s1
```

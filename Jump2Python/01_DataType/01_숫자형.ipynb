##### 정수형 (integer)은 말 그대로 정수를 뜻하는 자료형

```python
a = 123 
type(a)
```

##### 실수형 (floating-point)은 소수점이 포함된 숫자를 말한다.

```python
a = -3.45
type(a)
```

```python
# 아주 크거나 아주 작은 숫자를 표현할때는 e나 E를 사용해 표현할 수 있다.
a = 4.24e10
print(a, type(a))
```

##### 8진수와 16진수 : 8진수는 0o, 16진수는 0x로 표현한다.

```python
a = 0o177
print(a)

b = 0xABC
print(b)
```

##### 연산자

- 사칙 연산 : +, -, *, /

- x의 y제곱을 나타내는 ** 연산자

- 나눗셈 후 나머지를 반환하는 % 연산자

- 나눗셈 후 몫을 반환하는 // 연산자

- 복합 연산자 : =(대입 연산자), -=(감소된 만큼을 대입하는 대입연산자), +=(더해진 만큼을 대입하는 대입연산자)

```python
a = 3 
b = 4
print(a + b)
print(a / b)
print(a ** b)
print(7%3)
print(7//3)
a += 1
print(a)
```

```python
def add_many(*args):
    result = 0
    for i in args:
        result += i
    return result
```

```python
result = add_many(1, 2, 3, 4)
print(result)
```

```python
lst = [k for k in range(10)]
result = add_many(*lst)
print(result)
```

```python
def add_mul(choice, *args):
    if choice == 'add':
        result = 0
        for i in args:
            result += i
    elif choice =='mul':
        result = 1
        for i in args:
            result *= i
    return result
```

```python
add_mul('add', 1, 2, 3, 4, 5)
```

```python
add_mul('mul', 1, 2, 3, 4, 5)
```

```python
def print_kwargs(**kwargs):
    print(kwargs)
```

```python
print_kwargs(a = 1)
```

```python
print_kwargs(name = '홍길동', age = 25, city = '서울', job = '개발자')
```

```python
def create_profile(**info):
    print("=== 프로필 정보 ===")
    for key, value in info.items():
        print(f"{key} : {value}")
```

```python
create_profile(name = '김철수', age = 30, hobby = '독서', job = '프로그래머')
```

```python
def mixed_function(name, *args, **kwargs):
    print(f"이름 : {name}")
    print(f"추가 인수들 : {args}")
    print(f"키워드 인수들 : {kwargs}")
```

```python
mixed_function('홍길동', 1, 2, 3, age = 25, city = '서울', job = '개발자')
```

```python
def add_and_mul(a,b):
    return a + b, a * b

result1, result2 = add_and_mul(3, 4)
print(result1, result2)
```

```python
a = 1
def vartest(a):
    a += 1

vartest(a)
print(a)  # a의 값은 여전히 1, 함수 안에서 a를 변경했지만, 함수 밖의 a에는 영향을 주지 않음
```

##### 함수 안에서 함수 밖의 변수를 바꾸기

```python
a = 1
def vartest(a):
    a += 1
    return a
a = vartest(a)
print(a)
```

```python
a = 1
def vartest():
    global a
    a += 1

vartest()
print(a)
```

```python
add = lambda a, b : a + b
result = add(3, 4)
print(result)  
```

```python
def add(a, b):
    """
    두 숫자를 더하는 함수

    Parameters:
    a (int, float): 첫 번째 숫자
    b (int, float): 두 번째 숫자

    Returns:
    int, float: 두 숫자의 합
    """
    return a + b

# 독스트링 확인하기
print(add.__doc__)
```

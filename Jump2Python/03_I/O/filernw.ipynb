```python
f = open("새파일.txt", 'w')
for i in range(1, 11):
    data = f"{i}번째 줄입니다.\n"
    f.write(data)
f.close()
```

```python
f = open("새파일.txt", 'r')
line = f.readline()
print(line)
f.close()
```

```python
f = open("새파일.txt", 'r')
while True:
    line = f.readline()
    if not line: break
    print(line)
f.close()
```

```python
f = open("새파일.txt", 'r')
lines = f.readlines()
for line in lines:
    print(line)
f.close()
```

```python
f = open("새파일.txt", 'a') # 파일을 이어쓰기 모드로 열기
for i in range(11, 20):
    data = f"{i}번째 줄입니다.\n"
    f.write(data)
f.close()
```

```python
f = open("새파일.txt", 'r')
for line in f:
    print(line)
f.close()
```

```python
f = open('foo.txt', 'w')
f.write("Life is too short, you need python")
f.close()
```

```python
# file_with.py
with open("foo.txt", "w") as f: # with문을 사용하면 close()를 호출하지 않아도 자동으로 파일을 닫아줌
    f.write("Life is too short, you need python")
```

```python
f = open("foo.txt", 'r')
for line in f:
    print(line)
f.close()
```

```python
with open("foo.txt", "w") as f:
    f.write("Hello")
    print(f.closed)  # False (아직 열려 있음)

print(f.closed)  # True (with 블록을 벗어나 자동으로 닫힘)
```

```python
with open("test.txt", "w") as f:
    content = "Hello, Python!"
    f.write(content)

print(content)  # "Hello, Python!" 출력
```

```python
# 한글 파일 쓰기
with open("한글파일.txt", "w", encoding="utf-8") as f:
    f.write("안녕하세요, 파이썬!")

# 한글 파일 읽기
with open("한글파일.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)
```
